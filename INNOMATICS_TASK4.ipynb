```python
#Problem 1: Mobile Recharge Validation System

def recharge(amt):
    valid_plans=[199,299,399,599]
    
    if amt<= 50:
        print("Recharge must be at least ₹50")
    elif amt not in valid_plans:
        print("Invalid Plan selected")
    else:
        print("Recharge Successful")
        
while True:
    amt=int(input("Enter Recharge amount:"))
    if recharge(amt):
        break
```

```python
#Problem 2: Inventory Reorder Alert System

invt={"Laptop": 10,
    "Mouse": 25,
    "Keyboard": 12,
    "Monitor": 20}

def check_i(prod):
    for i in prod:
        if prod[i]<15:
            print(f"{i} Reorder Alert")
        else:
            print(f"{i} Stock OK")
        
check_i(invt)
```

```python
#Problem 3: Student Result Processing System

def Student(marks):
        sum=0
        for i in marks:
            sum += i
        avg=sum / len(marks)
        
        if avg>=50:
            print(f"Student PASS {avg}")
        else:
            print(f"Student FAIL {avg}")
marks=[10,20,48,90,89]
res=Student(marks)
m1=[10,20,5,6]
res1=Student(m1)
```

```python
def calculate_fare(distance, peak):
    
    fare = 50 + (12 * distance)
    if peak == 1:
        fare += fare * 0.25
    return fare


while True:
    distance = float(input("Enter distance in km: "))
    
    print("Peak Hour?")
    print("1. Yes")
    print("2. No")
    
    choice = int(input("Select option (1/2): "))

    total = calculate_fare(distance, choice)
    print("Total Fare = ₹", total)

    print("\nDo you want to calculate again?")
    print("1. Yes")
    print("2. No")
    
    retry = int(input("Select option (1/2): "))

    if retry == 2:
        break
```

```python
def cal_att(att):
    present_days = 0
    
    for i in att:
        if i == "P":
            present_days += 1
    
    per = (present_days / len(att)) * 100
    
    print(f"Attendance: {per}%")
    
    if per >= 75:
        return "Employee Eligible"
    else:
        return "Employee Not Eligible"


att = ["P", "A", "P", "P", "A", "P", "P"]
status = cal_att(att)
print("Status:", status)
```

```python
#Problem 6: Password Strength Checker

def pass_checker(passw):
    
    if len(passw) < 8:
        return "Minimum 8 characters required"
    
    has_digit = False
    has_special = False
    
    for ch in passw:
        if ch.isdigit():
            has_digit = True
        if ch in "@#$":
            has_special = True
    
    if has_digit and has_special:
        return "Strong Password "
    else:
        return "Weak Password Must include digit and @#$"


pwd = input("Enter password: ")
print(pass_checker(pwd))
```
