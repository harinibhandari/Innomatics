```python
# Problem 1: Social Media – Post Engagement Analyzer

def analyze_post(like):
    count=0
    for i in likes:
        count+=i
    if count>=1000:
        status="Viral Post"
    else:
        status="Normal Engagement"
    print("Total Likes:", count)
    print("Post Status:", status)


likes=[100,200,400,500,300,20,35]
analyze_post(likes)
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker

def rainfall_check(rainfall_data ,required):    
    sum=0
    for i in rainfall_data:
        sum+=i
    avg=sum/len(rainfall_data)
    
    if avg>=required:
        status="Adequate Rainfall"
    else:
        status="Inadequate Rainfall"
    print("Average Rainfall:", int(avg))
    print("Rainfall Status:",status)
rainfall_data = [70, 75, 68, 80]
required=70
rainfall_check(rainfall_data ,required)
```

```python
# Problem 4: Social Media – Duplicate Account Detection

def check_duplicates(un):
    
    if len(un) != len(set(un)):
        res = "Yes"
    else:
        res = "No"
    print("Duplicate Accounts Found:", res)
un = ["harini", "ravi", "meena", "harini"]
check_duplicates(un)
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter

def premium_crop_filter(prices):
    
    premium = []
    for i in prices:
        if i > 2000:
            premium.append(i)
    
    print("Premium Crops:", premium)


# Example usage
prices = [1500, 2500, 3200, 1800]
premium_crop_filter(prices)
```

```python
# Problem 7: System Monitoring – Application Health Checker

def health_appli(error):
    if error==0:
        print("Healthy")
    elif error<=5:
        status="Minor Issues"
    else:
        status="Critical Issues"
    print("Error Count:",error)
    print("System Status:",status)
        
error=int(input("Enter Error count:"))
health_appli(error)
```

```python
# Problem 8: Banking – Daily Transaction Limit Checker

def validate_txn(amount):
    daily_limit=50000
    
    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"
    print("Transaction Amount:", amount)
    print("Transaction Status:", status)
    
amount = int(input("Enter Transaction Amount: "))
validate_txn(amount)    
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System

def cal_atten(atten):
    
    total_days = len(atten)
    present_days = 0
    for day in atten:
        if day == 1:
            present_days += 1
    per=(present_days / total_days) * 100
    
    if per >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"
    
    print("Attendance Percentage:", per)
    print("Exam Eligibility:", status)

atten = [1,1,1,0,1,1,1,1]
cal_atten(atten)
```

```python
# Problem 10: Smart Electricity Bill Analyzer

def electricity_bill(units):
    if units <= 100:
        bill = units * 3      
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)    
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    if bill < 500:
        usage = "Low Usage"
        
    elif bill <= 1500:
        usage = "Moderate Usage"
        
    else:
        usage = "High Usage"
    
    return bill, usage
units = int(input("Enter units consumed: "))
total_bill, usage_status = electricity_bill(units)

print("Total Bill:", total_bill)
print("Usage Level:", usage_status)
```
