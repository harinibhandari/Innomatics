```python
#1:Smart Parking Lot Management System

def smart_parking_mang(capacity,vehicle_logs):
    entry=0
    for i in vehicle_logs:
        if i == "IN":
            entry +=1
        elif i == "OUT" and entry >0:
            entry -=1
    print("Currently Parked Vehicles:", entry)
    
    if entry > capacity:
        print("Parking Status: Full - Alert!")
    else:
        print("Parking Status: Available")
capacity= 50
vehicle_logs=["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
smart_parking_mang(capacity,vehicle_logs)
```

```python
#2:Online Food Delivery Time Estimator

def delivery_time(dis,weather,traffic_level):
    
    time=dis*5
    if traffic_level=="High":
        time +=15
    elif traffic_level == "Medium":
        time += 8
    if weather == "Rainy":
        time += 10
       
    print(f"Estimated Delivery Time:{time} minutes")
delivery_time(8,"Rainy","High")  
```

```python
# 3:Movie Theatre Seat Occupancy Analyzer


def seat_booking(seats,booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / seats) * 100
    
    print("Occupancy:", int(occupancy), "%")
    
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 70:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")

booked = [1] * 150
seat_booking(200, booked)
    
    
```

```python
# 4:Cloud Server Load Classification System

def server_load(CPU_Readings):
    count=0
    for i in CPU_Readings:
        count += i
        avg=count/len(CPU_Readings)
    print(f"Average CPU Load: {int(avg)} %") 
    
    if avg < 50:
        status="Normal"
    elif avg >= 50 or avg < 80:
        status="Warning"
    else:
        status="Critical"
    
    print(f"Server Status: {status}")
    
CPU_Readings=[45, 60, 70, 85, 90]
server_load(CPU_Readings)
```

```python
#5 Smart Classroom Resource Usage Monitor

def track_usage(resource_usage):
    
    overuse_limit = 8      
    overused = []       
    for i in resource_usage:
        if resource_usage[i] > overuse_limit:
            overused.append(i)
            
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

resource_usage = {"Projector": 6,"AC": 9,"Lights": 4}
track_usage(resource_usage)
```

```python
# 6: Online Event Registration Capacity Controller

def event_regi(cap, regi):
    
    if regi <= cap:
        confirmed = regi
        waitlisted = 0
        status = "Open"
    else:
        confirmed = cap
        waitlisted = regi - cap
        status = "Closed"
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


event_regi(100, 105)
```
