```python
#Problem 1: Smart Light Controller

class smart_light:
    def __init__(self,name):
        self.name=name
    
    def turn_on(self):
        print(self.name,"is ON")
    
    def turn_off(self):
        print(self.name,"is OFF")
        
        
    
l=smart_light("Bedroom Light")
l.turn_on()
    
    
```

```python
#Problem 2: Employee ID Card System


class Employee:
    
    def __init__(self,name,id,dept):
        self.name=name
        self.id=id
        self.dept=dept
    
    def display(self):
        print("INSIDE CLASS \nEmployee ID Card")
        print(f"Employee Name:{self.name}")
        print(f"Employee ID:{self.id}")
        print(f"Department:{self.dept}")
        
e=Employee("Rahul","EMP102","IT")
e.display()
print("\n OUSTIDE CLASS\nEmployee ID Card")
print(f"Employee Name:{e.name}")
print(f"Employee ID:{e.id}")
print(f"Department:{e.dept}")
```

```python
#Problem 3: Mobile Contact Record

class Mobile:
    
    def __init__(self,name,phno):
        self.name=name
        self.phno=phno
        
    def display(self):
        print("INSIDE CLASS \nContact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phno}")
m=Mobile("Anita",9876543210)
m.display()

print("\n OUSTIDE CLASS\nContact Saved")
print(f"Name: {m.name}")
print(f"Phone: {m.phno}")
```

```python
#Problem 4: Product Price Tag Generator

class Product:
    
    def __init__(self,name,price):
        self.name=name
        self.price=price
        
    def display(self):
        print("INSIDE CLASS")
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")
        
p=Product("Headphones",2499)
p.display()

print("\nOUTSIDE CLASS")
print(f"Product: {p.name}")
print(f"Price: ₹{p.price}")
```

```python
#Problem 5: Movie Rating Display System

class Movie:
    
    def __init__(self,name,rating):
        self.name=name
        self.rating=rating    
    def display(self):
        print("INSIDE CLASS")
        print(f"Movie: {self.name}")
        print(f"Rating: ₹{self.rating}/5")
        
m=Movie("Inception",4.8)
m.display()
print("\nOUTSIDE CLASS")
print(f"Movie: {m.name}")
print(f"Rating: ₹{m.rating}/5")
```

```python
# Problem 6: Delivery Address Manager

class Delivery:
    def __init__(self, customer_name, add):
        self.customer_name = customer_name
        self.add = add
    
    def display(self):
        print("INSIDE CLASS")
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.add)

d = Delivery("Suman", "Hyderabad")
d.display()

print("\nOUTSIDE CLASS")
print("Delivery Details")
print("Customer:", d.customer_name)
print("Address:", d.add)
```
